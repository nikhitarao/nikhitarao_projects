```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Load base tables
uscities = pd.read_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/uscities.csv')
auto_make_model = pd.read_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/AUTO_MAKE_MODEL.csv')

# Function to generate random date
def random_date(start, end):
    if start > end:
        return None
    return start + timedelta(days=random.randint(0, (end - start).days))

# Function to generate agent data for a given year
def generate_agents(year, num_agents):
    agents = []
    for i in range(num_agents):
        agent_id = f"AG{str(716260 + i).zfill(6)}"
        agent_origin_date = random_date(datetime(year, 1, 1), datetime(year, 12, 31))
        captive_prob = random.uniform(0.05, 0.15)
        independent_prob = 1 - captive_prob
        agent_type = np.random.choice(["CAPTIVE", "INDEPENDENT"], p=[captive_prob, independent_prob])
        if agent_type == "INDEPENDENT":
            agent_status = np.random.choice(["ACTIVE", "INACTIVE"], p=[0.65, 0.35])
        else:
            agent_status = np.random.choice(["ACTIVE", "INACTIVE", "LOST"], p=[0.85, 0.10, 0.05])
        agent_inactive_from = random_date(agent_origin_date + timedelta(days=18*30), datetime(2024, 12, 31)) if agent_status == "INACTIVE" else None
        agent_state = "Texas"
        agent_county = uscities[uscities['STATE'] == 'Texas']['COUNTY'].sample().values[0]
        agent_city = uscities[(uscities['STATE'] == 'Texas') & (uscities['COUNTY'] == agent_county)]['CITY'].sample().values[0]
        agent_category = np.random.choice(["SILVER", "GOLD", "PLATINUM"], p=[0.60, 0.25, 0.15])
        if agent_category == "SILVER":
            agent_commission_rate = round(random.uniform(5, 10), 2)
            agent_premium_discount_rate = round(random.uniform(5, 25), 2)
        elif agent_category == "GOLD":
            agent_commission_rate = round(random.uniform(11, 15), 2)
            agent_premium_discount_rate = round(random.uniform(26, 30), 2)
        else:
            agent_commission_rate = round(random.uniform(16, 20), 2)
            agent_premium_discount_rate = round(random.uniform(31, 45), 2)
        agent_loss_reason = np.random.choice(["Better Commissions from Competitor", "Lack of Operational/Claims Support", "Lack of Premium Support"], p=[0.55, 0.15, 0.30]) if agent_status == "LOST" else None
        agent_lost_to = np.random.choice(["Competitor A", "Competitor B", "Competitor C", "Competitor D"], p=[0.6374, 0.1177, 0.1177, 0.1272]) if agent_status == "LOST" else None

        agents.append([
            agent_id, agent_origin_date.strftime('%Y-%m-%d'), agent_type, agent_status, 
            agent_inactive_from.strftime('%Y-%m-%d') if agent_inactive_from else None, 
            agent_state, agent_county, agent_city, agent_category, 
            agent_commission_rate, agent_premium_discount_rate, 
            agent_loss_reason, agent_lost_to
        ])
    return agents

# Generate data for each year
base_year = 2015
num_records = 497
data = []

for year in range(base_year, 2025):
    data.extend(generate_agents(year, num_records))
    num_records = int(num_records * random.uniform(1.075, 1.15))

# Create DataFrame and save to CSV
columns = [
    "AGENT_ID", "AGENT_ORIGIN_DATE", "AGENT_TYPE", "AGENT_STATUS", 
    "AGENT_INACTIVE_FROM", "AGENT_STATE", "AGENT_COUNTY", "AGENT_CITY", 
    "AGENT_CATEGORY", "AGENT_COMMISSION_RATE", "AGENT_PREMIUM_DISCOUNT_RATE", 
    "AGENT_LOSS_REASON", "AGENT_LOST_TO"
]
df = pd.DataFrame(data, columns=columns)
df.to_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/AUTO_INSURANCE_AGENT_MASTER.csv', index=False)
print("Synthetic data generated and saved to AUTO_INSURANCE_AGENT_MASTER.csv")
```
