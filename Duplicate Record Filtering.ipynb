```python
import pandas as pd
from tqdm import tqdm

# Define file paths
base_path = "C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/"
file_paths = {
    "AUTO_INSURANCE_AGENT_MASTER": base_path + "AUTO_INSURANCE_AGENT_MASTER.csv",
    "AUTO_INSURANCE_POLICY_MASTER": base_path + "AUTO_INSURANCE_POLICY_MASTER.csv",
    "AUTO_INSURANCE_CLAIMS_MASTER": base_path + "AUTO_INSURANCE_CLAIMS_MASTER.csv",
    "AUTO_INSURANCE_QUOTE_CONVERSION_MASTER": base_path + "AUTO_INSURANCE_QUOTE_CONVERSION_MASTER.csv"
}

# Define columns to filter on
filter_columns = {
    "AUTO_INSURANCE_AGENT_MASTER": "AGENT_ID",
    "AUTO_INSURANCE_POLICY_MASTER": "POLICY_NUMBER",
    "AUTO_INSURANCE_CLAIMS_MASTER": "CLAIMS_NUMBER",
    "AUTO_INSURANCE_QUOTE_CONVERSION_MASTER": "APPLICATION_ID"
}

# Function to remove duplicates and save the file
def remove_duplicates(file_path, filter_column):
    # Read the CSV file
    df = pd.read_csv(file_path)
    
    # Drop duplicates based on the filter column
    df_unique = df.drop_duplicates(subset=[filter_column])
    
    # Save the unique records back to the CSV file
    df_unique.to_csv(file_path, index=False)

# Process each file with a progress bar
for table_name, file_path in tqdm(file_paths.items(), desc="Processing tables"):
    remove_duplicates(file_path, filter_columns[table_name])

print("Duplicate records removed and files overwritten successfully.")
```
