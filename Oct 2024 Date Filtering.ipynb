```python
import pandas as pd
import numpy as np

# Define file paths
base_path = "C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/"
agent_master_file = base_path + "AUTO_INSURANCE_AGENT_MASTER.csv"
policy_master_file = base_path + "AUTO_INSURANCE_POLICY_MASTER.csv"
claims_master_file = base_path + "AUTO_INSURANCE_CLAIMS_MASTER.csv"
quote_conversion_master_file = base_path + "AUTO_INSURANCE_QUOTE_CONVERSION_MASTER.csv"

# Define date ranges
date_range_start = "2015-01-01"
date_range_end = "2024-12-31"
filter_date_agent = "2024-09-30"
filter_date_policy_claims = "2024-09-30"
filter_date_quote_conversion = "2024-10-31"

# Load data
agent_master_df = pd.read_csv(agent_master_file)
policy_master_df = pd.read_csv(policy_master_file)
claims_master_df = pd.read_csv(claims_master_file)
quote_conversion_master_df = pd.read_csv(quote_conversion_master_file)

# Filter AUTO_INSURANCE_AGENT_MASTER
agent_master_df['AGENT_ORIGIN_DATE'] = pd.to_datetime(agent_master_df['AGENT_ORIGIN_DATE'])
agent_master_df['AGENT_INACTIVE_FROM'] = pd.to_datetime(agent_master_df['AGENT_INACTIVE_FROM'])

agent_master_filtered = agent_master_df[
    (agent_master_df['AGENT_ORIGIN_DATE'] >= date_range_start) &
    (agent_master_df['AGENT_ORIGIN_DATE'] <= filter_date_agent)
]

agent_master_filtered.loc[
    agent_master_filtered['AGENT_INACTIVE_FROM'] > filter_date_agent, 
    ['AGENT_STATUS', 'AGENT_INACTIVE_FROM', 'AGENT_LOSS_REASON', 'AGENT_LOST_TO']
] = np.nan

agent_master_filtered.loc[
    agent_master_filtered['AGENT_STATUS'].isin(['INACTIVE', 'LOST']) & 
    (agent_master_filtered['AGENT_INACTIVE_FROM'] > filter_date_agent), 
    'AGENT_STATUS'
] = np.nan

agent_master_filtered.to_csv(base_path + "AUTO_INSURANCE_AGENT_MASTER_OCT_2024.csv", index=False)

# Filter AUTO_INSURANCE_POLICY_MASTER
policy_master_df['POLICY_BIND_DATE'] = pd.to_datetime(policy_master_df['POLICY_BIND_DATE'])

policy_master_filtered = policy_master_df[
    (policy_master_df['POLICY_BIND_DATE'] >= date_range_start) &
    (policy_master_df['POLICY_BIND_DATE'] <= filter_date_policy_claims)
]

policy_master_filtered.to_csv(base_path + "AUTO_INSURANCE_POLICY_MASTER_OCT_2024.csv", index=False)

# Filter AUTO_INSURANCE_CLAIMS_MASTER
claims_master_df['INCIDENT_DATE'] = pd.to_datetime(claims_master_df['INCIDENT_DATE'])

claims_master_filtered = claims_master_df[
    (claims_master_df['INCIDENT_DATE'] >= date_range_start) &
    (claims_master_df['INCIDENT_DATE'] <= filter_date_policy_claims)
]

claims_master_filtered.to_csv(base_path + "AUTO_INSURANCE_CLAIMS_MASTER_OCT_2024.csv", index=False)

# Filter AUTO_INSURANCE_QUOTE_CONVERSION_MASTER
quote_conversion_master_df['APPLICATION_DATE'] = pd.to_datetime(quote_conversion_master_df['APPLICATION_DATE'])
quote_conversion_master_df['QUOTE_DATE'] = pd.to_datetime(quote_conversion_master_df['QUOTE_DATE'])
quote_conversion_master_df['QUOTE_ACCEPTANCE_DATE'] = pd.to_datetime(quote_conversion_master_df['QUOTE_ACCEPTANCE_DATE'])

quote_conversion_filtered = quote_conversion_master_df[
    (quote_conversion_master_df['APPLICATION_DATE'] >= date_range_start) &
    (quote_conversion_master_df['APPLICATION_DATE'] <= filter_date_quote_conversion)
]

quote_conversion_filtered.loc[
    quote_conversion_filtered['APPLICATION_DATE'].between("2024-10-01", filter_date_quote_conversion), 
    ['QUOTE_CONVERTED', 'POLICY_NUMBER', 'CUSTOMER_ID', 'POLICY_PREMIUM', 'POLICY_DEDUCTIBLE', 'POLICY_CSL', 'POLICY_BIND_DATE', 'QUOTE_DATE', 'QUOTE_ACCEPTANCE_DATE', 'LOSS_REASON']
] = np.nan

quote_conversion_filtered.loc[
    quote_conversion_filtered['QUOTE_DATE'] > filter_date_quote_conversion, 
    ['QUOTE_DATE', 'QUOTE_ACCEPTANCE_DATE']
] = np.nan

quote_conversion_filtered.to_csv(base_path + "AUTO_INSURANCE_QUOTE_CONVERSION_MASTER_OCT_2024.csv", index=False)

print("Data filtering and saving completed successfully.")
```
