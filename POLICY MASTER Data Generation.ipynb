```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Load the base tables
uscities = pd.read_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/uscities.csv')
auto_make_model = pd.read_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/AUTO_MAKE_MODEL.csv')
auto_insurance_agent_master = pd.read_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/AUTO_INSURANCE_AGENT_MASTER.csv')

# Initialize parameters
base_year = 2015
end_year = 2024
base_records = 55011
policy_number_start = 71626089
customer_id_start = 716260

# Function to generate random date within a range
def random_date(start, end):
    return start + timedelta(days=random.randint(0, (end - start).days))

# Function to generate policy premium
def calculate_policy_premium(base_premium, cover_premium, safety_premium, umbrella_premium):
    policy_premium = base_premium + cover_premium - safety_premium + umbrella_premium
    return max(policy_premium, 750)

# Initialize lists to store generated data
data = []

# Generate data for each year
for year in range(base_year, end_year + 1):
    if year == base_year:
        num_records = base_records
    else:
        num_records = int(num_records * (1 + random.uniform(0.075, 0.15)))

    for i in range(num_records):
        policy_number = f'P{policy_number_start + i:08d}'
        policy_bind_date = random_date(datetime(year, 1, 1), datetime(year, 12, 31)).strftime('%Y-%m-%d')
        auto_make = random.choice(auto_make_model['AUTO_MAKE'].unique())
        auto_model = random.choice(auto_make_model[auto_make_model['AUTO_MAKE'] == auto_make]['AUTO_MODEL'].unique())
        policy_csl = random.choices(['100/300', '250/500', '500/1000'], weights=[47, 38, 15])[0]
        policy_deductible = random.choices([250, 500, 1000, 2000], weights=[15, 45, 30, 10])[0]
        no_of_drivers = random.choices([1, 2, 3, 4], weights=[50, 30, 15, 5])[0]
        comprehensive_cover = random.choices(['YES', 'NO'], weights=[85.46, 14.54])[0]
        collision_cover = random.choices(['YES', 'NO'], weights=[67.88, 32.12])[0]
        uninsured_motorist = random.choices(['YES', 'NO'], weights=[43.67, 56.33])[0]
        medical_benefits = random.choices(['YES', 'NO'], weights=[23.89, 76.11])[0]
        pip_cover = random.choices(['YES', 'NO'], weights=[67.21, 32.79])[0]
        accident_forgiveness = random.choices(['YES', 'NO'], weights=[24.68, 75.32])[0]
        roadside_assistance = random.choices(['YES', 'NO'], weights=[74.32, 25.68])[0]
        loss_of_use = random.choices(['YES', 'NO'], weights=[29.13, 70.87])[0]
        anti_theft_device = random.choices(['YES', 'NO'], weights=[29.13, 70.87])[0]
        safety_score = None if random.random() < 0.78 else round(random.uniform(15, 97), 2)
        umbrella_limit = None if random.random() < 0.88 else random.choice([1000000, 2500000, 5000000, 7500000])
        
        base_premium = random.randint(750, 2500)
        cover_premium = sum(random.randint(50, 350) for cover in [comprehensive_cover, collision_cover,
                                                                 uninsured_motorist, medical_benefits,
                                                                 pip_cover, accident_forgiveness,
                                                                 roadside_assistance, anti_theft_device] if cover == 'YES')
        safety_premium = safety_score * 100 if safety_score else 0
        umbrella_premium = {1000000: 250, 2500000: 500, 5000000: 750, 7500000: 1000}.get(umbrella_limit, 0)
        
        policy_premium = calculate_policy_premium(base_premium, cover_premium, safety_premium, umbrella_premium)
        
        customer_id = f'CUST{customer_id_start + i:06d}'
        
        agent_id_row = auto_insurance_agent_master[auto_insurance_agent_master['AGENT_INACTIVE_FROM'].isnull() | 
                                                   (auto_insurance_agent_master['AGENT_INACTIVE_FROM'] > policy_bind_date)].sample()
        
        agent_id = agent_id_row['AGENT_ID'].values[0]
        policy_state = agent_id_row['AGENT_STATE'].values[0]
        policy_city = agent_id_row['AGENT_CITY'].values[0]
        policy_county = agent_id_row['AGENT_COUNTY'].values[0]

        data.append([policy_number,
                     policy_bind_date,
                     auto_make,
                     auto_model,
                     policy_csl,
                     policy_deductible,
                     no_of_drivers,
                     comprehensive_cover,
                     collision_cover,
                     uninsured_motorist,
                     medical_benefits,
                     pip_cover,
                     accident_forgiveness,
                     roadside_assistance,
                     loss_of_use,
                     anti_theft_device,
                     safety_score,
                     umbrella_limit,
                     policy_premium,
                     customer_id,
                     agent_id,
                     policy_state,
                     policy_city,
                     policy_county])

# Create DataFrame and save to CSV
columns = ['POLICY_NUMBER',
           'POLICY_BIND_DATE',
           'AUTO_MAKE',
           'AUTO_MODEL',
           'POLICY_CSL',
           'POLICY_DEDUCTIBLE',
           'NO_OF_DRIVERS',
           'COMPREHENSIVE_COVER',
           'COLLISION_COVER',
           'UNINSURED_MOTORIST',
           'MEDICAL_BENEFITS',
           'PIP_COVER',
           'ACCIDENT_FORGIVENESS',
           'ROADSIDE_ASSISTANCE',
           'LOSS_OF_USE',
           'ANTI_THEFT_DEVICE',
           'SAFETY_SCORE',
           'UMBRELLA_LIMIT',
           'POLICY_PREMIUM',
           'CUSTOMER_ID',
           'AGENT_ID',
           'POLICY_STATE',
           'POLICY_CITY',
           'POLICY_COUNTY']

df_auto_insurance_policy_master = pd.DataFrame(data, columns=columns)
df_auto_insurance_policy_master.to_csv('C:/Users/10741867/OneDrive - LTIMindtree/Documents/Datasets/Solution Datasets/Solution 1 - PNC Quote Conversion and Prioritization/Datasets/Regenerated Datasets/AUTO_INSURANCE_POLICY_MASTER.csv', index=False)
```
